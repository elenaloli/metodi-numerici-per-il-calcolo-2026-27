
### Introduzione al corso Metodi Numerici per il calcolo

**Analisi dei dati** L’analisi dei dati è l’insieme di metodi matematici, statistici e informatici utilizzati per estrarre informazioni utili da una collezione di dati. L’obiettivo non è semplicemente osservare dei numeri, ma trasformarli in conoscenza, individuando regolarità, relazioni, anomalie e tendenze.
I dati possono provenire da fonti molto diverse: misurazioni scientifiche, sensori, immagini, transazioni economiche, social network, esperimenti, questionari oppure applicazioni web. Possono inoltre essere numerici, testuali, categorici, temporali o visuali.

E i **metodi numerici**?

Molti problemi di analisi dei dati non possono essere risolti mediante formule esatte. È quindi necessario ricorrere a *metodi numerici*, ossia algoritmi che producono soluzioni approssimate attraverso una sequenza di operazioni matematiche.

I metodi numerici vengono utilizzati, ad esempio, per:
- risolvere sistemi di equazioni;
- approssimare dati mediante rette, curve o funzioni;
- calcolare minimi e massimi;
- ridurre la dimensionalità dei dati;
- classificare oggetti;
- ricostruire segnali e immagini;
- stimare i parametri di un modello;
- addestrare modelli di machine learning.

L’analisi dei dati è quindi un **processo interdisciplinare** nel quale **matematica, statistica e informatica** collaborano per rispondere a domande come:

- Quali caratteristiche emergono dai dati?
- Esiste una relazione tra due o più variabili?
- È possibile prevedere un valore futuro?
- Un’osservazione è normale oppure anomala?
- Come possiamo rappresentare molti dati utilizzando poche informazioni significative?

Nel corso di Metodi numerici per l’analisi dei dati studieremo gli strumenti matematici e computazionali che consentono di rispondere a queste domande, prestando particolare attenzione alla formulazione dei problemi, agli algoritmi utilizzati e all’affidabilità delle soluzioni ottenute.

Il corso è strutturato *per argomenti*, cioè affronteremo i diversi problemi matematici che sono alla base delle risposte alle domande precedenti separatamente.

Per ogni problema matematico considereremo alcuni aspetti:

1. Quando esiste ed è eventualmente unica la soluzione del problema nel continuo?
2. Quali algoritmi risolvono il problema nel discreto? Quanli sono le condizioni per potere applicare questi algoritmi?
3. Come posso implementare questi algoritmi in Python?
4. Esempi di applicazione degli algoritmi su problemi test e, in quaqlche caso particolare, su dati reali.



I macro-argomenti che verranno trattai in questo corso sia nel modulo I che nel modulo II sono i seguenti:

1. Elementi di base per il corso: calcolo multivariato (modulo I) e  numeri  ed aritmetica finiti (modulo II)

2. Metodi numerici per problemi non lineari: ricerca degli zeri (modulo II) e ottimizzazione numerica (modulo I)

3. Algebra lineare numerica: fattorizzazione di matrici per risolvere sistemi lineari con matrice quadrata (modulo II) e problemi di minimi quadrati (modulo I) 

4. Analisi dati. Modelli predittivi: interpolazione (modulo II) e regressione (modulo I). 

5. Equazioni differenziali ordinarie: problemi a valori iniziali (cenno) (modulo I)
